```python
coffee_clean = coffee[
    [
        "Location.Country",
        "Data.Scores.Aroma",
        "Data.Scores.Flavor",
        "Data.Scores.Acidity",
        "Data.Scores.Total"
    ]
]
coffee_clean = coffee_clean.dropna()
```

```python
coffee_clean.head()
```

```python
coffee_clean = coffee_clean.dropna()
coffee_clean.head()
```

```python
coffee_clean.shape
```

```python
for index, row in coffee_clean.iterrows():
    if row["Data.Scores.Total"] >= 85:
        print(row)
```

```python
hightotalscoreclean = coffee_clean[coffee_clean["Data.Scores.Total"] >= 85]
hightotalscoreclean.head(10)
```

```python
hightotalscoreclean.shape
```

```python
average_aroma = coffee_clean["Data.Scores.Aroma"].mean()
average_flavor = coffee_clean["Data.Scores.Flavor"].mean()
average_acidity = coffee_clean["Data.Scores.Acidity"].mean()
average_total = coffee_clean["Data.Scores.Total"].mean()

print("Average aroma score:", average_aroma)
print("Average flavor score:", average_flavor)
print("Average acidity score:", average_acidity)
print("Average total score:", average_total)
```

```python
correlations = coffee_clean[
    [
        "Data.Scores.Aroma",
        "Data.Scores.Flavor",
        "Data.Scores.Acidity",
        "Data.Scores.Total"
    ]
].corr()

correlations["Data.Scores.Total"]

print(correlations)
```

```python
plt.scatter(coffee_clean["Data.Scores.Flavor"], coffee_clean["Data.Scores.Total"])
plt.xlabel("Flavor Score")
plt.ylabel("Total Coffee Score")
plt.title("Flavor Score vs Total Coffee Score")
plt.savefig("flavor_vs_total.png")
plt.show()
```

```python
plt.scatter(coffee_clean["Data.Scores.Aroma"], coffee_clean["Data.Scores.Total"])
plt.xlabel("Aroma Score")
plt.ylabel("Total Coffee Score")
plt.title("Aroma Score vs Total Coffee Score")
plt.savefig("aroma_vs_total.png")
plt.show()
```

```python
plt.scatter(coffee_clean["Data.Scores.Acidity"], coffee_clean["Data.Scores.Total"])
plt.xlabel("Acidity Score")
plt.ylabel("Total Coffee Score")
plt.title("Acidity Score vs Total Coffee Score")
plt.savefig("acidity_vs_total.png")
plt.show()
```

```python
country_scores = coffee_clean.groupby("Location.Country")["Data.Scores.Total"].mean()

country_scores = country_scores.sort_values(ascending=False).head(10)

print(country_scores)
```

```python
country_scores.plot(kind="bar")
plt.xlabel("Country")
plt.ylabel("Average Total Coffee Score")
plt.title("Top 10 Countries by Average Coffee Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("top10_country_scores.png")
plt.show()
```
